# 12장 상속

 - 상속에 대한 이론과 실제 학습

```python
class Car:
    def __init__(self, maker, model, colar, price):
        self.maker = maker
        self.model = model
        self.colar = colar
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return f"Maker : {self.maker}, Model : {self.model}, Colar : {self.colar}, Price : {self.price}"

    def __str__(self):
        return f"Maker : {self.maker}, Model : {self.model}, Colar : {self.colar}, Price : {self.price}"



class ElectricCar(Car):
    def __init__(self, maker, model, colar,price, bat_size):
        super().__init__(maker, model, colar, price)
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self):
        self.batterSize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}"

    def getDesc(self):
        return self.__str__()

class Truck(Car):
    def __init__(self, maker, model, colar, price, payload):
        super().__init__(maker, model, colar, price)
        self.payload : payload

    def __str__(self):
        return f"{super().__str__()}, Payload : {self.payload}"

car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Model s", "red", 8000, 81)
print(ecar1)

truck1 = Truck("Ford", "deawoo","red", 16000, 100)
print(truck1)

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))

print(isinstance(truck1, Vehicle))
```
